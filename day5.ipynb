```python
#project 5
price1=0.30
price2=0.35
price3=0.40
limit1=100
limit2=300
monthly_fee=12

def calculate_bill(usage):
    """Return the monthly bill in Euros"""
    if usage<=limit1:
            bill=usage*price1+monthly_fee
    elif usage<=limit2:
            bill=limit1*price1+(usage-limit1)*price2+monthly_fee
    else:
            bill=limit1*price1+(limit2-limit1)*price2+(usage-limit2)*price3+monthly_fee
    return bill
print(calculate_bill(250))
import csv
months=[]
usages=[]
prices=[]
try:
    with open("energy_year.csv") as f:
        reader=csv.DictReader(f)
        for row in reader:
            try:
                usage=float(row["usage_kwh"])
                price=float(row["price_eur"])
            except ValueError:
                print(f"Warning: bad value in row {row}, skipped")
                continue
            months.append(row["month"])
            usages.append(usage)
            prices.append(price)
except FileNotFoundError:
    print("File not found. Check the file name and folder.")
bills = [calculate_bill(u) for u in usages]
total_usage = sum(usages)
total_bill = sum(bills)
highest = months[bills.index(max(bills))]
lowest = months[bills.index(min(bills))]

print(f"Total usage: {total_usage:.0f} kWh")
print(f"Total bill: {total_bill:.2f} EUR")
print(f"Highest bill: {highest}")
print(f"Lowest bill: {lowest}")
with open("bill_report.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["month", "usage_kwh", "bill_eur"])
    for m, u, b in zip(months, usages, bills):
        writer.writerow([m, u, round(b, 2)])

```
