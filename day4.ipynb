```python
#Project 4, Refactor your bill calculator
price1=0.30
price2=0.35
price3=0.40
limit1=100
limit2=300
monthly_fee=12

def calculate_bill(usage):
    """Return the monthly bill in Euros"""
    if usage<=limit1:
            bill=usage*price1+monthly_fee
    elif usage<=limit2:
            bill=limit1*price1+(usage-limit1)*price2+monthly_fee
    else:
            bill=limit1*price1+(limit2-limit1)*price2+(usage-limit2)*price3+monthly_fee
    return bill

def yearly_summary(usages):
    """Return total, average bill and index of highest month."""
    bills=[]
    for u in usages:
        bill=calculate_bill(u)
        bills.append(bill)
    total=sum(bills)
    average=total/len(bills)
    highest=bills.index(max(bills))
    return total,average,highest
print(calculate_bill(250))
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
usage = [320, 290, 260, 210, 180, 150, 140, 145, 190, 240, 290, 340]

total, average, highest_index = yearly_summary(usage)
print(f"Total: {total:.2f} EUR")
print(f"Average monthly bill: {average:.2f} EUR")
print(f"Highest bill in: {months[highest_index]}")
```
