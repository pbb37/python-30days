```python
#project 3

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
usage = [320, 290, 260, 210, 180, 150, 140, 145, 190, 240, 290, 340]
total=sum(usage)
average=sum(usage)/len(usage)
print(f"Total: {total} kWh, average: {average:.2f} kWh")
highest_value=max(usage)
highest_month=months[usage.index(highest_value)]
lowest_value=min(usage)
lowest_month=months[usage.index(lowest_value)]
print(f"Highest value is {highest_value} in {highest_month}")
print(f"Lowest value is {lowest_value} in {lowest_month}")
for m,u in zip(months,usage):
    if u>average:
        print(m,u)
usage_by_month=dict(zip(months,usage))
print(usage_by_month)
winter_total=usage_by_month["Dec"]+usage_by_month["Jan"]+usage_by_month["Feb"]
print(winter_total)
top3 = sorted(usage, reverse=True)[:3]
print(top3)
```
