```python
#project Simulate Hourly Solar Generation for a Day
import numpy as np
PEAK_POWER_KW=4
SUNRISE=6
SUNSET=18
hours=np.arange(0,24)
angle= np.pi*(hours-SUNRISE)/(SUNSET-SUNRISE)
power=(np.sin(angle)*PEAK_POWER_KW)
daylight = (hours >= SUNRISE) & (hours <= SUNSET)
power = np.where(daylight, power, 0)
energy_per_hour=power*1
daily_energy=energy_per_hour.sum()
np.random.seed(42)
cloud_factor=np.random.uniform(0.7,1.0,size=24)
power_with_cloud=power*cloud_factor
print (f"{power.max()} KW at hour {power.argmax()}")
print(f"Hours with power above 0- {(power>0).sum()}")
print(f"Clear sky energy {energy_per_hour.sum()}")
print(f"with clouds energy {power_with_cloud.sum()}")
capacity_factor= daily_energy / (PEAK_POWER_KW * 24) * 100
print(f"Capacity factor {capacity_factor}")

import matplotlib.pyplot as plt

plt.plot(hours, power, label="Clear sky")
plt.plot(hours, power_with_cloud, label="With clouds")
plt.xlabel("Hour of day")
plt.ylabel("Power (kW)")
plt.title("Simulated solar generation")
plt.grid(True)
plt.legend()
plt.savefig("solar_day.png", dpi=150)
plt.show()
```
