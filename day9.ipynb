```python
#miniproject
import pandas as pd
df=pd.read_csv("energy_year.csv")
print(df.shape)
df["co2_kg"]=df["usage_kwh"]*0.38
df["cost_eur"]=df["usage_kwh"]*df["price_eur"]
df["season"]="Autumn"
winter=df["month"].isin(["Dec","Jan","Feb"])
df.loc[winter,"season"]="winter"
spring=df["month"].isin(["Mar","Apr","May"])
df.loc[spring,"season"]="spring"
summer=df["month"].isin(["Jun","Jul","Aug"])
df.loc[summer,"season"]="summer"
#print(df[["month", "season"]])
season_usage=df.groupby("season")["usage_kwh"].sum()
print(season_usage)
top=df.loc[df["usage_kwh"].idxmax(), "month"]  #highest
low=df.loc[df["usage_kwh"].idxmin(), "month"] #lowest
above = df[df["usage_kwh"] > df["usage_kwh"].mean()]
print(len(above))
top_price=df.loc[df["cost_eur"].idxmax(), "month"]

print(df[df["usage_kwh"] > df["usage_kwh"].mean()])
print(top)
print(top_price)
df.to_csv("energy_report.csv", index=False)
```
