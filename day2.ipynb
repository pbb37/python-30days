```python
#project electricity bill calculator for 12 months
price1=0.30
price2=0.35
price3=0.40
limit1=100
limit2=300
monthly_fee=12
total=0
max_bill=0
max_month=0
for month in range(1,13):
        usage=float(input(f"Kwh used in this month {month}:"))
        while usage<0:
            print("check the value")
            usage=float(input(f"Kwh used in this month{month}:"))
        if usage<=limit1:
            bill=usage*price1+monthly_fee
        elif usage<=limit2:
            bill=limit1*price1+(usage-limit1)*price2+monthly_fee
        else:
            bill=limit1*price1+(limit2-limit1)*price2+(usage-limit2)*price3+monthly_fee
        total+=bill
        if bill>max_bill:
            max_bill=bill
            max_month=month
        print(f"month {month}:{bill:.2f}Euro")
print(f"total {total:.2f}Euros")
print(f"Highest bill {max_bill:.2f}Euros and in month {max_month}")


print(f"Average bill{total/12:.2f} Euros")


```
