```python
def read_numbers(filename):
    """Read one number per line and return them as a list."""
    numbers=[]
    try:
        with open(filename) as f:
            for line in f:
                numbers.append(float(line.strip()))
    except FileNotFoundError:
        print(f"File {filename} not found")
    return numbers

print(read_numbers("numbers.txt"))
print(read_numbers("missing.txt"))
```

```python
#miniproject
import csv

ELECTRICITY_FACTOR = 0.38   # kg CO2 per kWh
GAS_FACTOR = 0.20           # kg CO2 per kWh
CAR_FACTOR = 0.17           # kg CO2 per km
class Household:
    def __init__(self, name, people, electricity_kwh, gas_kwh, car_km):
        self.name=name
        self.people=people
        self.electricity_kwh=electricity_kwh
        self.gas_kwh=gas_kwh
        self.car_km=car_km
    def footprint(self):
        """Return dictionary of emissions in kg of co2"""
        electricity=self.electricity_kwh*ELECTRICITY_FACTOR
        gas=self.gas_kwh*GAS_FACTOR
        car=self.car_km*CAR_FACTOR
        total=electricity+gas+car
        return{"electricity": electricity,"gas": gas,"car": car,"total": total}
    def per_person(self):
        """Return emission per person"""
        return self.footprint()["total"]/self.people
    def __str__(self):
        return f"{self.name}:{self.per_person():.0f} kg per person"
def load_households(filename):
    """Read the CSV and return a list of Household objects."""
    households = []
    try:
        with open(filename) as f:
            reader=csv.DictReader(f)
            for row in reader:
                try:
                    people=int(row["people"])
                    electricity=float(row["electricity_kwh"])
                    gas=float(row["gas_kwh"])
                    car=float(row["car_km"])
                except ValueError:
                    print(f"Warning: bad value in row {row}, skipped")
                    continue
                if people < 1:
                    print(f"Warning: people must be at least 1 in {row['name']}, skipped")
                    continue
                households.append(Household(row["name"], people, electricity, gas, car))
    except FileNotFoundError:
        print(f"File {filename} not found")
    return households
households=load_households("households.csv")
if not households:
    print("No Data to Analyse")
else:
    total_all=0
    max_value=0
    max_name=""
    for h in households:
        fp = h.footprint()
        print(h)
        print(f"  electricity {fp['electricity']:.0f}, gas {fp['gas']:.0f}, "
              f"car {fp['car']:.0f}, total {fp['total']:.0f} kg")

        total_all += fp["total"]
        if h.per_person() > max_value:
            max_value = h.per_person()
            max_name = h.name

    print(f"\nHighest per person: {max_name} ({max_value:.0f} kg)")
    print(f"Total of all households: {total_all:.0f} kg")

  
    with open("footprint_report.csv", "w", newline="") as f:
        writer = csv.writer(f)
        writer.writerow(["name", "electricity_kg", "gas_kg", "car_kg", "total_kg", "per_person_kg"])
        for h in households:
            fp = h.footprint()
            writer.writerow([h.name, round(fp["electricity"]), round(fp["gas"]),
                             round(fp["car"]), round(fp["total"]), round(h.per_person())])
```
