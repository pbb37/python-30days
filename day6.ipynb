```python
#project
class SolarPanel:
    def __init__(self, name, power, efficiency):
        self.name = name
        self.power = power
        self.efficiency = efficiency

    def yearly_output(self, sun_hours=1000):
        """Return yearly energy in kWh"""
        return self.power * sun_hours * self.efficiency

    def co2_saved(self, factor=0.38):
        """Calculates kg of CO2 avoided per year"""
        return self.yearly_output() * factor

    def __str__(self):
        return f"SolarPanel({self.name}, {self.power} kW, {self.efficiency})"


class Household:
    def __init__(self, name, yearly_usage_kwh):
        self.name = name
        self.yearly_usage_kwh = yearly_usage_kwh
        self.panels = []

    def add_panel(self, panel):
        """Appends a panel to the household's panel list."""
        self.panels.append(panel)

    def total_generation(self):
        """Returns the sum of all panel outputs in kWh."""
        return sum(panel.yearly_output() for panel in self.panels)

    def self_sufficiency(self):
        """Returns generation divided by usage as a percentage, protected against division by zero."""
        try:
            return (self.total_generation() / self.yearly_usage_kwh) * 100
        except ZeroDivisionError:
            return 0.0


if __name__ == "__main__":
    # Household with 3500 kWh usage
    my_household = Household(name="Family Home", yearly_usage_kwh=3500)

    # Use 'power' instead of 'power_kw' to match __init__
    panel_1 = SolarPanel(name="Panel 1", power=5, efficiency=0.85)
    panel_2 = SolarPanel(name="Panel 2", power=3, efficiency=0.80)

    my_household.add_panel(panel_1)
    my_household.add_panel(panel_2)

    total_gen = my_household.total_generation()
    total_co2 = sum(panel.co2_saved() for panel in my_household.panels)
    self_suff = my_household.self_sufficiency()

    print(f"Panel 1 Output: {panel_1.yearly_output()} kWh")
    print(f"Panel 2 Output: {panel_2.yearly_output()} kWh")
    print(f"Total Generation: {total_gen} kWh")
    print(f"CO2 Saved by both: {total_co2:.0f} kg")
    print(f"Self-sufficiency: {self_suff:.1f} %")
```
